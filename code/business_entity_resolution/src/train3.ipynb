```python
block_counts = (
    df.groupby(["country", "name_core"])
      .size()
)

print("Total blocks:", len(block_counts))

print("\nLargest blocks:")
print(block_counts.sort_values(ascending=False).head(20))

print("\nBlocks containing multiple records:",
      (block_counts > 1).sum())
```

```python
strong_blocks = (
    df.groupby(
        ["country", "name_core", "address_normalized"],
        dropna=False
    ).size()
)

print("Total blocks:", len(strong_blocks))

print("\nBlocks containing multiple records:",
      (strong_blocks > 1).sum())

print("\nLargest blocks:")
print(strong_blocks.sort_values(ascending=False).head(20))
```

```python
exact_blocks = strong_blocks[strong_blocks > 1]

print("Rows belonging to multi-record blocks:",
      exact_blocks.sum())

print("Number of multi-record blocks:",
      len(exact_blocks))

print("\nBlock size distribution:")
print(exact_blocks.value_counts().sort_index())
```
